```python
import numpy as np
```

```python
arr=np.array([2,3,4,5,6,7,7,8,8,9,10,10,11,11,12,12,13,13,14,14,14,14,201])
arr
```

# to remove the outliers we have to calculate q1 and q3 IQR

```python
Q1=np.percentile(arr,25)
Q1
```

```python
Q3=np.percentile(arr,75)
Q3
```

```python
IQR=Q3-Q1
IQR
```

# upper fence and lower fence

```python
UF=Q3+(1.5*IQR)
UF
```

```python
LF=Q1-(1.5*IQR)
LF
```

# remove outliers

```python
l=[]
for i in arr:
    if i>=LF and i<=UF:
        l.append(i)
        arr2=np.array(l)
```

### before removing the outlier


```python
print(arr)
```

### after removing the outliers

```python
print(arr2)
```
